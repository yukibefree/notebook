```python
# データ読み込み用の関数をインポート
import sys, os
from pathlib import Path
sys.path.append(os.pardir)  # 親ディレクトリのファイルをインポートするための設定
from src.ensemble import ensemble_submissions, stacking_ensemble, blending_ensemble

# データディレクトリ
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / 'data'
SUBMISSION = DATA_DIR / 'submission'
ENSEMBLE = SUBMISSION / 'ensemble'

os.makedirs(ENSEMBLE, exist_ok=True)
print(ENSEMBLE)
```

```python
# アンサンブルの対象ファイル
ensemble_files = [
    SUBMISSION / 'submission_lightgbm_20250625_224651.csv',
    SUBMISSION / 'submit_svr_pca_v6.csv',
    SUBMISSION / 'submission_nn_20250630_003938.csv',
    SUBMISSION / 'submission_random_forest.csv',
]
```

```python
# 平均によるアンサンブル
ensemble_submissions(
    ensemble_files,
    output_path=ENSEMBLE,
    filename='submission_lgbm_svr_pca_nn_rf_mean',
    method='mean'
)
```

```python
# 加重平均によるアンサンブル
ensemble_submissions(
    ensemble_files,
    output_path=ENSEMBLE,
    filename='submission_lgbm_svr_pca_nn_rf_weighted',
    method='weighted',
    weights=[
        0.2,
        0.75,
        0.02,
        0.025
    ]
)
```

```python
# スタッキングによるアンサンブル
stacking_ensemble(
    ensemble_files,
    output_path=ENSEMBLE,
    filename='submission_lgbm_svr_pca_stacking'
)
```

```python
# ブレンディングによるアンサンブル
# TODO: 学習予測結果の保存
blending_ensemble(
    val_pred_files=[
        'val_pred_lgbm.csv',
        'val_pred_svr.csv',
        'val_pred_rf.csv'
    ],
    val_true_file='val_true.csv',
    test_pred_files=ensemble_files,
    output_path=ENSEMBLE,
    filename='submission_lgbm_svr_pca_blending'
)
```
