```python
# データ読み込み用の関数をインポート
import sys, os
from pathlib import Path
sys.path.append(os.pardir)  # 親ディレクトリのファイルをインポートするための設定
from src.ensemble import ensemble_submissions, stacking_ensemble, blending_ensemble
from datetime import datetime

# データディレクトリ
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / 'data'
SUBMISSION = DATA_DIR / 'submission'
ENSEMBLE = SUBMISSION / 'ensemble'

os.makedirs(ENSEMBLE, exist_ok=True)
print(ENSEMBLE)
```
